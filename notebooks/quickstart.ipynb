### 🚀 Step 2: Run Live Cloud Extraction (`unitbytes/craigslist-scraper`)
Grab your free Apify API token from **[Apify Console](https://apify.com/unitbytes/craigslist-scraper?fpr=939u3w&fp_sid=gh_classifieds_colab)** (includes `$5.00` free monthly usage credit) and run the cell below. `getpass` prompts for your token securely so it is never stored in notebook outputs:


```python
!pip install -q apify-client pandas

from getpass import getpass
import pandas as pd
from apify_client import ApifyClient

APIFY_TOKEN = getpass("Enter your Apify API Token (from https://apify.com/unitbytes/craigslist-scraper?fpr=939u3w&fp_sid=gh_classifieds_colab): ").strip()
if not APIFY_TOKEN:
    raise ValueError("Please enter a valid Apify API token to run live extraction.")

client = ApifyClient(APIFY_TOKEN)

run_input = {
    "searchQuery": "macbook pro",
    "category": "sss",
    "subdomain": "losangeles",
    "scrapeDetails": True,
    "hasImage": False,
    "postedToday": False
}

print("Running unitbytes/craigslist-scraper...")
run = client.actor("unitbytes/craigslist-scraper").call(run_input=run_input)
items = list(client.dataset(run["defaultDatasetId"]).iterate_items())

df_live = pd.DataFrame(items)
print(f"Extracted {len(df_live)} live records!")
df_live.to_csv("craigslist-scraper-results.csv", index=False)
df_live.head(10)
```
